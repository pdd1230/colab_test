````python
a = [10, 20, 30]                  # list
b = (10, 20, 30)                  # tuple
c = {10, 20, 30}                  # set
d = {"name": "kim", "age": 20}    # dict

print(type(a))
print(type(b))
print(type(c))
print(type(d))
````

````python
import numpy as np

a = np.array([10, 20, 30])
# a = np.array([[10, 20, 30], [40, 50, 60]])
# ndarray → 수치 계산을 위한 배열
print(a)
print(type(a))
````

````python
import numpy as np
import pandas as pd

a = [10, 20, 30]   # 리스트
b = np.array(a)    # ndarray 배열
s = pd.Series(b)   #

# print(a)
# print(b)
print(s)
print()
print(type(a))   # <class 'list'>
print(type(b))   # <class 'numpy.ndarray'>
print(type(s))   # <class 'pandas.Series'>
````

````python
import numpy as np
import pandas as pd

a =[[10, 20, 30], [40,50,60]]   # 리스트
b = np.array(a)    # ndarray 배열
s = pd.DataFrame(b)   #

print(a)
print(b)
print(s)
print()
print(type(a))   # <class 'list'>
print(type(b))   # <class 'numpy.ndarray'>
print(type(s))   # <class 'pandas.DataFrame'>
````

````python
import torch

x = torch.tensor([10, 20, 30])
print(x)
print(type(x))
# Tensor
# → 딥러닝 계산에서 사용하는 다차원 데이터
````

````python
# numpy 배열의 연산

arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[7,8,9],[10,11,12]])

# 사칙연산
print('numpy 배열 더하기 = \n', arr1 + arr2)
print('numpy 배열 빼기 = \n', arr2 - arr1)
print('numpy 배열 곱하기 = \n', arr1 * arr2)
print('numpy 배열 나누기 = \n', arr2 / arr1)
````
